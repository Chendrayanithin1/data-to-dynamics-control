Implementing SVD from Scractch, First Step in understand the Core Foundational Mathematics of Linear Algebra for the Dynamic Systems.

```python
import numpy as np
import svd
```

Creating a 6 x 4 non-square matrix with random values

```python
A = np.random.rand(6, 4)
```

Finding the SVD using the Built in function to compare the results with the implemented function

```python
# Compute the SVD using built-in function
U,S,Vt = np.linalg.svd(A, full_matrices=False)
```

Finding the SVD using the function built from scratch

```python
U_t,S_t, Vt_t = svd.compute_svd(A)
```

Comparing the difference between inbuilt computation and scratch built function

```python
np.round( np.abs(U) - np.abs(U_t), 3)
```

```python
np.round( np.abs(Vt) - np.abs(Vt_t), 3)
```

```python
np.round( np.abs(S) - np.abs(S_t), 3)
```
